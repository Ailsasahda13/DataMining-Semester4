```python
# Import library yang diperlukan
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
from google.colab import drive
drive.mount('/content/drive')

# Membaca file dari Google Drive
df = pd.read_csv(
    '/content/drive/MyDrive/dataset/credit_data_sample.csv',
    sep=';'
)

# Menampilkan 5 baris pertama
print(df.head())
```

```python
features = ['Age', 'Credit_amount', 'Duration',
'Employment_since']
X = df[features]
```

```python
# Standarisasi data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Elbow Method
inertia = []
K = range(1, 10)

# Scale the data first
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Plot hasil elbow
plt.figure(figsize=(8,5))
plt.plot(K, inertia, 'bx-')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Inertia')
plt.title('Metode Elbow untuk Menentukan k')
plt.grid(True)
plt.show()
```

```python
# Buat model KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)
# Lihat hasil clustering
print(df.head())
```

```python
# Visualisasi hasil clustering
plt.figure(figsize=(8,6))
sns.scatterplot(x=df['Credit_amount'], y=df['Duration'],
hue=df['Cluster'], palette='viridis')
plt.title('Cluster Calon Kreditur Berdasarkan Jumlah Kredit dan Durasi')
plt.xlabel('Credit Amount')
plt.ylabel('Duration')
plt.grid(True)
plt.show()
```

```python
# 7a. Hitung Silhouette Score
silhouette_avg = silhouette_score(X_scaled,
df['Cluster'])
print(f"\nSilhouette Score: {silhouette_avg:.3f}")

# 7b. Tampilkan inertia akhir
print(f"Inertia (k=3): {kmeans.inertia_:.3f}")

# Step 8 — Visualisasi hasil clustering (2 fitur: Credit_amount vs Duration)
plt.figure(figsize=(8,6))
sns.scatterplot(
  x=df['Credit_amount'],
  y=df['Duration'],
  hue=df['Cluster'],
  palette='viridis',
  s=100
)
plt.title('Cluster Calon Kreditur Berdasarkan Credit Amount dan Duration')
plt.xlabel('Credit Amount')
plt.ylabel('Duration')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```
